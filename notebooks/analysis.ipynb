```python
import laspy
import pandas as pd
import numpy as np

# load las data
las_BART = laspy.read("data/BART_cropped.laz")
las_OSBS = laspy.read("data/OSBS_cropped.laz")
las_TALL = laspy.read("data/TALL_cropped.laz")

# convert to dataframe 
las_OSBS_df = pd.DataFrame({
    "X": np.asarray(las_OSBS.x),
    "Y": np.asarray(las_OSBS.y),
    "Z": np.asarray(las_OSBS.z),
    "Classification": las_OSBS.classification,
    "Intensity": las_OSBS.intensity,
    "red": las_OSBS.red,
    "green": las_OSBS.green,
    "blue": las_OSBS.blue,
    "Return_number": np.asarray(las_OSBS.return_number)
})

las_TALL_df = pd.DataFrame({
    "X": np.asarray(las_TALL.x),
    "Y": np.asarray(las_TALL.y),
    "Z": np.asarray(las_TALL.z),
    "Classification": las_TALL.classification,
    "Intensity": las_TALL.intensity,
    "red": las_TALL.red,
    "green": las_TALL.green,
    "blue": las_TALL.blue,
    "Return_number": np.asarray(las_TALL.return_number)
})


las_BART_df = pd.DataFrame({
    "X": np.asarray(las_BART.x),
    "Y": np.asarray(las_BART.y),
    "Z": np.asarray(las_BART.z),
    "Classification": las_BART.classification,
    "Intensity": las_BART.intensity,
    "red": las_BART.red,
    "green": las_BART.green,
    "blue": las_BART.blue,
    "Return_number": np.asarray(las_BART.return_number)
})

las_BART_df.head()


```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.kdeplot(data = las_BART_df, y = np.array(las_BART_df['Z']), color = 'black', bw=0.1)
plt.axhline(y = las_BART_df["Z"].quantile(0.98), color='red', linestyle='--', label = 'RH98')
plt.axhline(y = las_BART_df["Z"].quantile(0.50), color='blue', linestyle='--', label = 'RH50')
plt.axhline(y = las_BART_df["Z"].quantile(0.25), color='green', linestyle='--', label = 'RH25')
plt.ylabel("Height (m)")
plt.title("Relative Heights for Old Growth Deciduous Forest")
plt.legend()
plt.show()

sns.kdeplot(data = las_OSBS_df, y = np.array(las_OSBS_df['Z']), color = 'black', bw=0.1)
plt.axhline(y = las_OSBS_df["Z"].quantile(0.98), color='red', linestyle='--', label = 'RH98')
plt.axhline(y = las_OSBS_df["Z"].quantile(0.50), color='blue', linestyle='--', label = 'RH50')
plt.axhline(y = las_OSBS_df["Z"].quantile(0.25), color='green', linestyle='--', label = 'RH25')
plt.ylabel("Height (m)")
plt.title("Relative Heights for Longleaf Pine Savanna")
plt.legend()
plt.show()

sns.kdeplot(data = las_TALL_df, y = np.array(las_TALL_df['Z']), color = 'black', bw=0.1)
plt.axhline(y = las_TALL_df["Z"].quantile(0.98), color='red', linestyle='--', label = 'RH98')
plt.axhline(y = las_TALL_df["Z"].quantile(0.50), color='blue', linestyle='--', label = 'RH50')
plt.axhline(y = las_TALL_df["Z"].quantile(0.25), color='green', linestyle='--', label = 'RH25')
plt.ylabel("Height (m)")
plt.title("Relative Heights for Mixed Pine Forest")
plt.legend()
plt.show()
```
